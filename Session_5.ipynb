## Coding 1 — Visualizing Amplitude Interference

In this experiment, we will observe how quantum amplitudes change as a qubit passes through multiple quantum gates. We will use the **statevector** to examine the quantum state before and after applying two Hadamard gates.

The goal is to see how amplitudes can **reinforce or cancel each other**, producing interference and changing the final measurement probabilities.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a single-qubit circuit
qc = QuantumCircuit(1)

# Apply the first Hadamard gate
qc.h(0)

# Inspect the state after the first H
state_after_h = Statevector.from_instruction(qc)

print("State after first H:")
print(state_after_h)

# Apply the second Hadamard gate
qc.h(0)

# Inspect the final state
state_after_hh = Statevector.from_instruction(qc)

print("\nState after second H:")
print(state_after_hh)
```

## Coding 2 — Phase Kickback

In this experiment, we will explore **phase kickback**, where the action of a controlled quantum operation can introduce a phase onto another qubit. We will use a two-qubit circuit and inspect the statevector to see how the phase changes without immediately changing the measurement probabilities.

The goal is to understand how **phase information can be transferred and later converted into observable differences through interference**.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a two-qubit circuit
qc = QuantumCircuit(2)

# Prepare both qubits in |+⟩
qc.h(0)
qc.h(1)

# Apply a controlled-Z gate
qc.cz(0, 1)

# Inspect the statevector
state = Statevector.from_instruction(qc)

print("Statevector after phase kickback:")
print(state)
```

## Coding 3 — Interference-Based State Discrimination

The states |+⟩ and |−⟩ have the same measurement probabilities in the computational basis, but they contain different relative phases.

In this experiment, we will use a Hadamard gate to convert this hidden phase difference into a measurable difference. This demonstrates how **interference can reveal information about a quantum state that is not visible from direct measurement**.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a circuit for |+⟩
plus_circuit = QuantumCircuit(1)
plus_circuit.h(0)

# Create a circuit for |−⟩
minus_circuit = QuantumCircuit(1)
minus_circuit.x(0)
minus_circuit.h(0)

# Inspect the initial states
plus_state = Statevector.from_instruction(plus_circuit)
minus_state = Statevector.from_instruction(minus_circuit)

print("State |+⟩:")
print(plus_state)

print("\nState |−⟩:")
print(minus_state)
```

## Coding 4 — Amplitude Amplification: 2-Qubit Demo

In this experiment, we will see how quantum interference can be used to **increase the probability of a desired state**.

We begin with an equal superposition of the four two-qubit states and use an oracle to mark `|11⟩`. We then apply the Grover diffusion operator, which amplifies the amplitude of the marked state.

This demonstrates the basic mechanism of **amplitude amplification**, which forms the foundation of Grover's search algorithm.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a 2-qubit circuit
qc = QuantumCircuit(2)

# Step 1: Create equal superposition
qc.h(0)
qc.h(1)

# Inspect the state before the oracle
state_before_oracle = Statevector.from_instruction(qc)

print("Before oracle:")
print(state_before_oracle)

# Step 2: Oracle - mark |11⟩
qc.cz(0, 1)

# Inspect the state after the oracle
state_after_oracle = Statevector.from_instruction(qc)

print("\nAfter oracle:")
print(state_after_oracle)

# Step 3: Grover diffusion operator
qc.h(0)
qc.h(1)

qc.x(0)
qc.x(1)

qc.h(1)
qc.cx(0, 1)
qc.h(1)

qc.x(0)
qc.x(1)

qc.h(0)
qc.h(1)

# Inspect the final state
final_state = Statevector.from_instruction(qc)

print("\nAfter amplitude amplification:")
print(final_state)
```

## Coding 5 — Oracle + Diffusion

In this experiment, we will separate the two main components of amplitude amplification: the **oracle** and the **diffusion operator**.

The oracle marks a target state by changing its phase, while the diffusion operator uses interference to amplify the marked state's amplitude.

This modular approach shows how these two operations work together and forms the basic structure of **Grover's search algorithm**.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create the oracle
def oracle():
    oracle_circuit = QuantumCircuit(2)

    # Mark |11⟩ by applying a phase flip
    oracle_circuit.cz(0, 1)

    return oracle_circuit


# Create the diffusion operator
def diffusion():
    diffusion_circuit = QuantumCircuit(2)

    # Transform to the computational basis
    diffusion_circuit.h(0)
    diffusion_circuit.h(1)

    # Flip |00⟩
    diffusion_circuit.x(0)
    diffusion_circuit.x(1)

    # Apply controlled phase flip
    diffusion_circuit.h(1)
    diffusion_circuit.cx(0, 1)
    diffusion_circuit.h(1)

    # Undo the X operations
    diffusion_circuit.x(0)
    diffusion_circuit.x(1)

    # Return to the original basis
    diffusion_circuit.h(0)
    diffusion_circuit.h(1)

    return diffusion_circuit


# Create the main circuit
qc = QuantumCircuit(2)

# Create equal superposition
qc.h(0)
qc.h(1)

print("Initial circuit:")
print(qc)

# Apply the oracle
qc.compose(oracle(), inplace=True)

print("\nAfter oracle:")
print(qc)

# Inspect state after oracle
state_after_oracle = Statevector.from_instruction(qc)
print("\nState after oracle:")
print(state_after_oracle)

# Apply diffusion
qc.compose(diffusion(), inplace=True)

print("\nAfter diffusion:")
print(qc)

# Inspect final state
final_state = Statevector.from_instruction(qc)
print("\nFinal state:")
print(final_state)
```
